```python
def gera_dados(num_registros= 600):
  """Gera um DataFrame do Pandas com dados de vendas Fictícios."""

  print(f"\nIniciando a geração de {num_registros} registros de vendas...")
  produtos = {
    "Teclado Mecânico": {"categoria": "Periféricos", "preco": 299.90},
    "Mouse Vertical": {"categoria": "Periféricos", "preco": 149.90},
    "Mousepad extra grande": {"categoria": "Acessórios", "preco": 79.90},
    "Headset gamer": {"categoria": "Áudio", "preco": 249.90},
    "Monitor gamer 144 Hz": {"categoria": "Monitores", "preco": 1199.90},
    "Cadeira gamer": {"categoria": "Móveis", "preco": 999.90},
    "Controle sem fio": {"categoria": "Periféricos", "preco": 349.90},
    "Webcam Full HD": {"categoria": "Streaming", "preco": 199.90},
    "Microfone USB": {"categoria": "Áudio", "preco": 329.90},
    "Suporte para headset": {"categoria": "Acessórios", "preco": 59.90},
    "Placa de vídeo": {"categoria": "Hardware", "preco": 2499.90},
    "Memória RAM 16 GB": {"categoria": "Hardware", "preco": 279.90},
    "SSD NVMe 1 TB": {"categoria": "Armazenamento", "preco": 449.90},
    "Gabinete com iluminação RGB": {"categoria": "Gabinetes", "preco": 399.90},
    "Kit de ventoinhas RGB": {"categoria": "Refrigeração", "preco": 129.90},
}
  lista_produtos = list(produtos.keys())

  cidade_estado = {
    "Acre": "AC", "Alagoas": "AL", "Amapá": "AP","Amazonas": "AM", 
    "Bahia": "BA", "Ceará": "CE", "Distrito Federal": "DF", "Espírito Santo": "ES",
}
  lista_cidades = list(cidade_estado.keys())

  dados_vendas = []

  data_inicial = datetime(2026, 1, 1)

  for i in range(num_registros):
    produto_nome = random.choice(lista_produtos)
    cidade = random.choice(lista_cidades)

    quantidade = np.random.randint(1, 8)

    data_pedido = data_inicial + timedelta(days = int(i/5), hours = random.randint(0, 23))  
    if produto_nome in ['Mouse Vertical', 'Teclado Mecânico']:
      preco_unitario = produtos[produto_nome]['preco'] * np.random.uniform(0.9, 1.0)
    else:
      preco_unitario = produtos[produto_nome]['preco']

    dados_vendas.append({
      'ID_pedido': 1000 + i + 1,
      'Data_pedido': data_pedido,
      'Nome_produto':produto_nome,
      'Categoria': produtos[produto_nome]['categoria'],
      'Preco_unitario': round(preco_unitario, 2),
      'Quanditdade': quantidade,
      'Id_cliente': np.random.randint(100, 10000),
      'Cidade': cidade,
      'Estado': cidade_estado[cidade]
    })

  print("Geração de dados concluída.\n")

  return pd.DataFrame(dados_vendas)
```

```python
df_vendas = gera_dados(3000)
```
